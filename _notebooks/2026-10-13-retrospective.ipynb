## Personal Growth & Ground Zero (Work Habits Included)

I grew a lot personally. I gained some new skills, and I also learned a crucial lesson: the easy way is not the best way.

- Learned how to make effortless and clean pull requests.
- Learned to work within a system without scrap CSS and garbage code.
- Learned to prompt AI within a system's requirements and overall improved my prompting game. As a whole with AI, by prompting a lot, I learned what works and what starts to get a better result.
- Learned more parts of the system and started to understand the un-explained parts of the repo.

## MCQ Retrospective

I rushed through the test and didn't read through the questions. I chose the easy way, so I did not gain knowledge. Afterward, I took my time going over a couple of questions in depth with my teammates. I now know better.

## Team Improvements

As a team, Raymond, Pranay, and I have grown closer, and we now know how to split tasks evenly. Pranay and Raymond worked on connecting the website's Flask backend and frontend, while I used that connection, along with the correct OCS grammar and pages grammar, to make the trial mockup of the website. Then, using your input, Raymond and I condensed the website and worked on Login Page v1 and the Admin Panel. After that, we ran into issues, so all of us as a team researched different parts of the workflow and prompted AI to finally get the login and custom admin panel made within the workflow and integration of the repo.
