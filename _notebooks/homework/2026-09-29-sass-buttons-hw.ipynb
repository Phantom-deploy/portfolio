## Homework Hack #1: Adaptive Resource Toolbar

The three resource badges are grouped in an `ocs__links` container, and each one uses the tone that matches how urgent its category is:

| Category | Classes | Why |
| --- | --- | --- |
| Urgent Support | `ocs__btn alert-red fill pill` | Red and filled, so it stands out the most |
| Self-Care | `ocs__btn alert-green fill pill` | Green for a positive, supportive action |
| Community | `ocs__btn accent outline pill` | Outlined, so it is visible but lower priority |

## Homework Hack #2: Primary vs. Secondary Action Priority

The check-in panel becomes one action row inside `ocs__links`, where each button's style shows how important it is:

| Action | Priority | Classes |
| --- | --- | --- |
| Log Mood Entry | Primary | `ocs__btn accent fill` |
| View Weekly Trends | Secondary | `ocs__btn outline` |
| Crisis Hotline | Alert | `ocs__btn alert-red outline` |

"Log Mood Entry" is a `<button>` because it performs an action on the page. The other two are `<a>` links because they go to another page.

{% capture challenge5 %}
Homework Hack 2: a filled primary action, an outline secondary action, and a red outline alert action in ocs__links. Press Run.
{% endcapture %}

{% capture code5 %}
// Edit the HTML inside the quotes, then press Run.
