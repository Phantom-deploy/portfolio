## Popcorn Hack: Find the Bug

The toggle's track `<span>` was missing its class. Adding `ocs__toggle-track` brings back the visible switch, and `aria-hidden="true"` tells screen readers to skip it because it is only for looks.

## Homework: Study Settings Panel

A settings panel with three toggles. Each one has its own label and controls a different part of the page:

| Toggle | What it does when it is on |
| --- | --- |
| Show Notifications | Shows a callout with new lesson updates |
| Show Study Tips | Shows a card with study tips |
| Highlight Due Dates | Turns the due date cell into an accent cell |

Under the toggles, a status line shows how many toggles are on, and it updates every time one is flipped. Comments in the JavaScript explain how each part works.

// Grab every toggle input inside this runner. querySelectorAll gives a list,
// so the same code works no matter how many toggles there are.
const toggles = outputElement.querySelectorAll('.ocs__toggle-input');
const countOutput = outputElement.querySelector('#toggle-count');

// Each feature is its own small function, so the toggle IDs stay connected
// to exactly one thing on the page.
function showNotifications(isOn) {
  // The hidden attribute hides an element without any extra styling.
  outputElement.querySelector('#notify-box').hidden = !isOn;
}

function showTips(isOn) {
  outputElement.querySelector('#tips-box').hidden = !isOn;
}

function highlightDueDate(isOn) {
  // classList.toggle adds the OCS accent class when isOn is true
  // and removes it when isOn is false.
  outputElement.querySelector('#due-cell').classList.toggle('ocs__grid-cell--accent', isOn);
}

const features = {
  'notify-toggle': showNotifications,
  'tips-toggle': showTips,
  'due-toggle': highlightDueDate
};

function updatePage() {
  let onCount = 0;
  toggles.forEach(function (toggle) {
    // .checked is true when the toggle is on and false when it is off.
    // This is selection: IF the toggle is on, apply the feature, ELSE undo it.
    features[toggle.id](toggle.checked);
    if (toggle.checked) {
      onCount++;
    }
  });
  // The count is rebuilt from scratch every time, so it can never drift
  // out of sync with what the toggles actually show.
  countOutput.textContent = 'Toggles on: ' + onCount + ' of ' + toggles.length;
}

// The "change" event fires every time a toggle is flipped.
toggles.forEach(function (toggle) {
  toggle.addEventListener('change', updatePage);
});

updatePage();
