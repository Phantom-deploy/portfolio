## Homework Hack

The "How to Pat a Cat" guide is rebuilt with semantic tags inside an OCS card:

* Title: `<h2 class="ocs__section-title">`
* Section headings: `<h3>`
* Body text: `<p class="ocs__text">`
* Important text: `<strong>`
* The three steps: one `<ol>` with three `<li>`

{% capture challenge2 %}
Homework Hack: the cat guide with semantic tags, an ordered list, strong emphasis, and OCS classes. Press Run.
{% endcapture %}

{% capture code2 %}
// Edit the HTML inside the quotes, then press Run.
outputElement.innerHTML = [
  '<div class="ocs__card">',
  '  <h2 class="ocs__section-title">How to Pat a Cat</h2>',
  '  <h3>Have you ever wanted to…</h3>',
  '  <p class="ocs__text">pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>',
  '  <p class="ocs__text"><strong>How to pat a cat:</strong></p>',
  '  <ol>',
  '    <li>put out your hand</li>',
  '    <li>touch the cat</li>',
  '    <li>move your hand back and forth</li>',
  '  </ol>',
  '  <h3>Congrats! You pat a cat :D</h3>',
  '</div>'
].join('');
{% endcapture %}
