## Homework

My 3.1 star counter, refactored into one list. Each element of `level_stars` is one level's collected stars, in level order: level one, level two, level three. It adds 5 to the second level, appends 7 for a new fourth level, then uses Example C's traversal to total the stars.

**Prediction:** `[12, 23, 10, 7]`, `4`, `52`

```python
# CODE_RUNNER: UESL 3.2 Homework - Grow my game
level_stars = [12, 18, 10]
if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5
level_stars.append(7)
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars
print(level_stars)
print(len(level_stars))
print(total_stars)
```

**Actual output:** `[12, 23, 10, 7]`, `4`, `52`

**What the list saves me:** with separate variables, every new level needs a new variable name and an edit to the line that adds up the total. With 100 levels, that is 100 variables and a sum line with 100 terms, and forgetting one gives a wrong total. With a list, a new level is one `append`, and the same loop totals any number of levels without being changed.

**Result:** `0`, then `0`. The loop runs zero times, so `total_stars` stays at its starting value of `0`, and checking the length first keeps `level_stars[1]` from causing an `IndexError`.

## Design Thinking

* **Player and maker need:** a maker wants to add levels without rewriting code, and players want to know which level earned which stars.
* **Goal:** store every level in one collection that grows easily and still totals correctly.
* **Representation I considered:** separate variables (like 3.1) compared with one list. I chose the list because adding a level no longer means changing the total.
* **Prototype:** the list, update, append, and traversal above.
* **Test:** the main test gave `52` for four levels, and the empty test gave `0`, `0` without an error.

**Clear level labels:** players should not have to understand indices. A matching list of names shows each level by name instead:

```python
# CODE_RUNNER: UESL 3.2 Design - Show levels by name, not index
level_names = ["Forest", "Cavern", "Summit", "Comet Cave"]
level_stars = [12, 23, 10, 7]
for i in range(len(level_stars)):
    print(level_names[i] + ": " + str(level_stars[i]) + " stars")
```
