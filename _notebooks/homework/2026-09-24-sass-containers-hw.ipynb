* One `ocs__container` holding one `ocs__card`
* An `ocs__section-title` and a one-line `ocs__description`
* A `cols-3` standard grid with a `--header` cell, an `--accent` cell, and a `--muted` cell
* A table with `ocs__table-wrap` + `ocs__table`, made of a header row and two body rows
* One `ocs__callout` line at the end
