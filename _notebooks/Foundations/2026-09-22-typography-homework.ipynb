## OCS Semantic HTML and Typography Grammar

**Homework: Refactor to Semantic Tags**

**Task from the lesson:** take the starter block, remove every custom class, and use semantic HTML tags instead (`h1` to `h4`, `p`, `ul` or `ol` with `li`, `strong`, `em`) so the Aesthetihawk SASS theme styles it. Keep the `%%html` line and the `UI_RUNNER` comment.

**How I solved it:** every custom class was doing a job that a real tag already does. I swapped each one for the tag that means the same thing, so the theme can style it and a screen reader can read it correctly.

| Starter code | Semantic tag | Why |
| --- | --- | --- |
| `<p class="title-font">` | `<h1>` | It is the page title, so it is the top heading |
| `<p class="section-font">` | `<h2>` | It names a section under the title |
| `<p class="body-text">` | `<p>` | Normal paragraph text |
| `<span class="very-important">` | `<strong>` | Marks important words, not just bold looks |
| Three `<p class="list-item">` lines numbered by hand | `<ol>` with three `<li>` | The order matters, so the browser should number it |

```python
%%html

<!-- UI_RUNNER: Typography Homework | show_code: true -->

<h1>Project Features</h1>

<h2>User Accounts</h2>

<p>Users can make an account and log in. This is a <strong>crucial</strong> feature.</p>

<p>Steps to register:</p>

<ol>
  <li>Click register</li>
  <li>Enter email</li>
  <li>Set password</li>
</ol>
```

### Output and What It Shows

The theme now styles everything on its own. The heading sizes come from `h1` and `h2`, the numbers come from the `ol`, and the important word is bold because of `strong`. There is not one class or inline style left in the block.

| Emphasis semantics | 0.1 | `very-important` became `strong` |
