## Two Things to Know Before You Start

**On asking the user questions.** Real programs use `prompt()` in JavaScript and `input()` in Python.
Both of those stop and wait for a person, which freezes a code runner in a web page. So the cells below
use a small helper called `ask()` that reads from a list of pretend answers. The logic is identical.
When you move this code into your own project, swap it back:

- JavaScript: `let drink = prompt("What drink would you like?");`
- Python: `drink = input("What drink would you like? ")`

**On counting list positions.** College Board pseudocode counts positions starting at **1**. JavaScript and
Python both start at **0**. Topics 9, 11, and 12 keep a counter that starts at 1 so the logic matches the
pseudocode, then subtract 1 when reaching into the real array. Look for the `- 1` and you will see exactly
where the two systems disagree.

## 2. Input

**The idea in this one:** A coffee shop order ticket.

Input lets a person hand data to your program. Pseudocode uses `INPUT()`. The `ask()` helper below stands in for `prompt()` and `input()` so the cell never freezes waiting on you.

**Why it matters for the exam:** Input is required for the Create Performance Task. It proves your program reacts to different data instead of doing the same thing every single time.

### JavaScript

```python
%%js

// CODE_RUNNER: Add a question that asks for the customer name, then print it on the ticket.

let answers = ["Latte", "2"];

function ask(question) {
  let reply = answers.shift();
  console.log(question + " " + reply);
  return reply;
}

let drink = ask("What drink would you like?");
let shots = ask("How many espresso shots?");

console.log("Order placed: " + drink);
console.log("Espresso shots: " + shots);
```

### Python

```python
# CODE_RUNNER: Add a question that asks for the customer name, then print it on the ticket.

answers = ["Latte", "2"]

def ask(question):
    reply = answers.pop(0)
    print(question + " " + reply)
    return reply

drink = ask("What drink would you like?")
shots = ask("How many espresso shots?")

print("Order placed: " + drink)
print("Espresso shots: " + shots)
```

## 3. List and Traversal

**The idea in this one:** Adding up elevation gain on a hiking trail.

A list holds many values under one name. A traversal walks through every item one at a time. Here each number is a climb in feet, and the loop stacks them into one total.

**Why it matters for the exam:** Using a list to manage data earns points for handling complexity. Exam questions love asking you to trace a loop that builds a running total.

### JavaScript

```python
%%js

// CODE_RUNNER: Add two more climbs to the list and check that the total goes up.

let climbs = [120, 340, 85, 210];
let totalGain = 0;

for (let climb of climbs) {
  totalGain = totalGain + climb;
}

console.log("Total elevation gain: " + totalGain + " feet");
```

### Python

```python
# CODE_RUNNER: Add two more climbs to the list and check that the total goes up.

climbs = [120, 340, 85, 210]
total_gain = 0

for climb in climbs:
    total_gain = total_gain + climb

print("Total elevation gain: " + str(total_gain) + " feet")
```

## 4. Procedure with a Parameter and a Return Value

**The idea in this one:** Turning an earthquake magnitude into a damage level.

A procedure is a named block of code you can call over and over. It takes a parameter as input and hands back an answer with a return. This one takes a number off the Richter scale and returns a word.

**Why it matters for the exam:** You must write at least one procedure of your own that takes a parameter and gets called at least once. This is the abstraction point on the scoring guide.

### JavaScript

```python
%%js

// CODE_RUNNER: Add a fourth level called Severe for anything 8 or above.

function getQuakeLevel(magnitude) {
  if (magnitude >= 7) {
    return "Major";
  } else {
    if (magnitude >= 5) {
      return "Moderate";
    } else {
      return "Minor";
    }
  }
}

let reading = 6.2;
let level = getQuakeLevel(reading);
console.log("Magnitude " + reading + " is a " + level + " quake");
```

### Python

```python
# CODE_RUNNER: Add a fourth level called Severe for anything 8 or above.

def get_quake_level(magnitude):
    if magnitude >= 7:
        return "Major"
    else:
        if magnitude >= 5:
            return "Moderate"
        else:
            return "Minor"

reading = 6.2
level = get_quake_level(reading)
print("Magnitude " + str(reading) + " is a " + level + " quake")
```

## 6. Selection

**The idea in this one:** Turning an air quality index number into a health advisory.

Selection means the program picks one path based on a true or false test. Only one branch runs. Notice that a low number is good here, which is the opposite of most score examples you have seen.

**Why it matters for the exam:** Selection is required in your algorithm. Exam questions ask which branch runs for a given value. Trace the conditions in order and stop at the very first one that is true.

### JavaScript

```python
%%js

// CODE_RUNNER: Change the reading to 165 and predict the advisory before you press Run.

let answers = ["120"];
function ask(q) { let r = answers.shift(); console.log(q + " " + r); return r; }

let aqi = Number(ask("Air quality index reading:"));
let advisory = "Unknown";

if (aqi <= 50) {
  advisory = "Good, go outside";
} else {
  if (aqi <= 100) {
    advisory = "Moderate, most people are fine";
  } else {
    if (aqi <= 150) {
      advisory = "Unhealthy for sensitive groups";
    } else {
      if (aqi <= 200) {
        advisory = "Unhealthy, limit time outdoors";
      } else {
        advisory = "Hazardous, stay inside";
      }
    }
  }
}

console.log("Advisory: " + advisory);
```

### Python

```python
# CODE_RUNNER: Change the reading to 165 and predict the advisory before you press Run.

answers = ["120"]

def ask(question):
    reply = answers.pop(0)
    print(question + " " + reply)
    return reply

aqi = int(ask("Air quality index reading:"))
advisory = "Unknown"

if aqi <= 50:
    advisory = "Good, go outside"
else:
    if aqi <= 100:
        advisory = "Moderate, most people are fine"
    else:
        if aqi <= 150:
            advisory = "Unhealthy for sensitive groups"
        else:
            if aqi <= 200:
                advisory = "Unhealthy, limit time outdoors"
            else:
                advisory = "Hazardous, stay inside"

print("Advisory: " + advisory)
```

## 7. Iteration with a Sentinel Loop

**The idea in this one:** A grocery self checkout scanner.

Iteration means repeating a block of code. A sentinel loop keeps going until the shopper types a stop word, which here is the word done. Pseudocode writes this as `REPEAT UNTIL`. Nobody knows ahead of time how many items will get scanned, and that is the whole point.

**Why it matters for the exam:** Iteration is required in your algorithm. Exam questions ask how many times the loop body runs, so count the passes and watch what the sentinel is.

### JavaScript

```python
%%js

// CODE_RUNNER: Add one more item to the answers list and run it again.

let answers = ["Bananas", "1.75", "scan", "Bread", "3.25", "scan", "Milk", "4.50", "done"];
function ask(q) { let r = answers.shift(); console.log(q + " " + r); return r; }

console.log("Self Checkout");

let cart = [];
let itemCount = 0;
let cartTotal = 0;
let nextAction = "scan";

while (nextAction !== "done") {
  let item = ask("Scan an item:");
  let price = Number(ask("Price:"));

  cart.push(item);
  itemCount = itemCount + 1;
  cartTotal = cartTotal + price;

  nextAction = ask("Scan another or type done:");
}

console.log("Cart: " + cart);
console.log("Items scanned: " + itemCount);
console.log("Total due: " + cartTotal + " dollars");
```

### Python

```python
# CODE_RUNNER: Add one more item to the answers list and run it again.

answers = ["Bananas", "1.75", "scan", "Bread", "3.25", "scan", "Milk", "4.50", "done"]

def ask(question):
    reply = answers.pop(0)
    print(question + " " + reply)
    return reply

print("Self Checkout")

cart = []
item_count = 0
cart_total = 0
next_action = "scan"

while next_action != "done":
    item = ask("Scan an item:")
    price = float(ask("Price:"))

    cart.append(item)
    item_count = item_count + 1
    cart_total = cart_total + price

    next_action = ask("Scan another or type done:")

print("Cart: " + str(cart))
print("Items scanned: " + str(item_count))
print("Total due: " + str(cart_total) + " dollars")
```

## 8. Algorithm, All Three Pieces Together

**The idea in this one:** A rainfall sensor deciding if there is a drought.

An algorithm puts sequence, selection, and iteration together to solve one real problem. This one loops through a week of sensor readings, guards against an empty list so it never divides by zero, and then judges the result.

**Why it matters for the exam:** This is where you earn the algorithm implementation point. One loop on its own is not enough, and one if statement on its own is not enough. The grader wants all three working together.

### JavaScript

```python
%%js

// CODE_RUNNER: Empty the readings list and run it. The guard should return 0 instead of crashing.

function averageRainfall(readings) {
  let total = 0;
  let count = 0;

  for (let drop of readings) {
    total = total + drop;
    count = count + 1;
  }

  if (count > 0) {
    let average = total / count;
    return average;
  } else {
    return 0;
  }
}

let weekReadings = [0.4, 1.2, 0.0, 0.8, 0.3];
let result = averageRainfall(weekReadings);

console.log("Readings: " + weekReadings);
console.log("Average rainfall: " + result + " inches per day");

if (result >= 1.0) {
  console.log("Status: Normal");
} else {
  console.log("Status: Drought watch");
}
```

### Python

```python
# CODE_RUNNER: Empty the readings list and run it. The guard should return 0 instead of crashing.

def average_rainfall(readings):
    total = 0
    count = 0

    for drop in readings:
        total = total + drop
        count = count + 1

    if count > 0:
        average = total / count
        return average
    else:
        return 0

week_readings = [0.4, 1.2, 0.0, 0.8, 0.3]
result = average_rainfall(week_readings)

print("Readings: " + str(week_readings))
print("Average rainfall: " + str(result) + " inches per day")

if result >= 1.0:
    print("Status: Normal")
else:
    print("Status: Drought watch")
```

## 9. Linear Search

**The idea in this one:** Finding a book barcode on a library shelf.

A linear search checks each item in order until it finds a match. If it runs off the end without a match it returns -1, which is the normal way for code to say not found. The counter starts at 1 so the position it reports matches pseudocode.

**Why it matters for the exam:** Linear search shows up on the exam constantly. Know what it returns when the item is missing, and know that in the worst case it has to check every single item.

### JavaScript

```python
%%js

// CODE_RUNNER: Search for a barcode that is not on the shelf and check that you get -1 back.

function findBook(shelf, target) {
  let index = 1;

  for (let code of shelf) {
    if (code === target) {
      return index;
    }
    index = index + 1;
  }

  return -1;
}

let shelf = ["978-0143", "978-0307", "978-0451", "978-0679"];

let lookingFor = "978-0451";
let position = findBook(shelf, lookingFor);

if (position > 0) {
  console.log("Barcode " + lookingFor + " is at shelf position " + position);
} else {
  console.log("That book is not on this shelf");
}
```

### Python

```python
# CODE_RUNNER: Search for a barcode that is not on the shelf and check that you get -1 back.

def find_book(shelf, target):
    index = 1

    for code in shelf:
        if code == target:
            return index
        index = index + 1

    return -1

shelf = ["978-0143", "978-0307", "978-0451", "978-0679"]

looking_for = "978-0451"
position = find_book(shelf, looking_for)

if position > 0:
    print("Barcode " + looking_for + " is at shelf position " + str(position))
else:
    print("That book is not on this shelf")
```

## 10. List Operations

**The idea in this one:** Managing an airport boarding queue.

These are the four list operations the exam expects you to know: add to the end, insert at a position, remove at a position, and get the length. Watch closely, because pseudocode position 2 is real index 1.

**Why it matters for the exam:** Exam questions trace how these operations shift everything else. Removing an item slides every later item down by one, and that shift is a favorite trap.

### JavaScript

```python
%%js

// CODE_RUNNER: Insert a group at position 1 and watch every other position slide over.

let boardingQueue = ["Group A", "Group B"];
console.log("Initial: " + boardingQueue);

boardingQueue.push("Group C");
console.log("After APPEND: " + boardingQueue);

boardingQueue.splice(1, 0, "Priority");   // pseudocode INSERT at position 2
console.log("After INSERT at 2: " + boardingQueue);

boardingQueue.splice(2, 1);               // pseudocode REMOVE at position 3
console.log("After REMOVE at 3: " + boardingQueue);

let waiting = boardingQueue.length;
console.log("Groups still waiting: " + waiting);

for (let group of boardingQueue) {
  console.log("Now boarding: " + group);
}
```

### Python

```python
# CODE_RUNNER: Insert a group at position 1 and watch every other position slide over.

boarding_queue = ["Group A", "Group B"]
print("Initial: " + str(boarding_queue))

boarding_queue.append("Group C")
print("After APPEND: " + str(boarding_queue))

boarding_queue.insert(1, "Priority")      # pseudocode INSERT at position 2
print("After INSERT at 2: " + str(boarding_queue))

boarding_queue.pop(2)                     # pseudocode REMOVE at position 3
print("After REMOVE at 3: " + str(boarding_queue))

waiting = len(boarding_queue)
print("Groups still waiting: " + str(waiting))

for group in boarding_queue:
    print("Now boarding: " + group)
```

## 11. Parallel Lists, Find the Position First

**The idea in this one:** Looking up a chemical element by its symbol on the periodic table.

Two lists are parallel when position 1 of the first belongs with position 1 of the second. Here the symbols and the full names line up. The procedure searches the symbols, hands back a position, and then the calling code uses that position to reach into the names list.

**Why it matters for the exam:** Parallel lists let one search answer a question about a completely different list. That is real data handling and it counts toward managing complexity.

### JavaScript

```python
%%js

// CODE_RUNNER: Look up He instead and confirm the element name that comes back.

function findSymbol(symbolList, target) {
  let index = 1;

  for (let symbol of symbolList) {
    if (symbol === target) {
      return index;
    }
    index = index + 1;
  }

  return -1;
}

let symbols = ["H", "He", "Li", "Be"];
let names   = ["Hydrogen", "Helium", "Lithium", "Beryllium"];

let lookingFor = "Li";
let position = findSymbol(symbols, lookingFor);

if (position > 0) {
  let elementName = names[position - 1];   // pseudocode starts at 1, JavaScript starts at 0
  console.log("Symbol found at position: " + position);
  console.log("Element: " + elementName);
} else {
  console.log("That symbol is not in this table");
}
```

### Python

```python
# CODE_RUNNER: Look up He instead and confirm the element name that comes back.

def find_symbol(symbol_list, target):
    index = 1

    for symbol in symbol_list:
        if symbol == target:
            return index
        index = index + 1

    return -1

symbols = ["H", "He", "Li", "Be"]
names   = ["Hydrogen", "Helium", "Lithium", "Beryllium"]

looking_for = "Li"
position = find_symbol(symbols, looking_for)

if position > 0:
    element_name = names[position - 1]     # pseudocode starts at 1, Python starts at 0
    print("Symbol found at position: " + str(position))
    print("Element: " + element_name)
else:
    print("That symbol is not in this table")
```

## 12. Parallel Lists, Return the Answer Directly

**The idea in this one:** A vending machine turning a slot code into a price.

This is the same kind of search as topic 11, with one real improvement. The procedure does the position math on the inside and hands back the price itself. The code that calls it never has to know that positions exist at all. Put topic 11 and topic 12 next to each other and the difference is easy to see.

**Why it matters for the exam:** Hiding the messy part inside a procedure is exactly what abstraction means. This comparison gives you something concrete to write about in your written response.

### JavaScript

```python
%%js

// CODE_RUNNER: Add a third parallel list for the snack name and return that instead of the price.

function findPrice(slots, prices, targetSlot) {
  let index = 1;

  for (let slot of slots) {
    if (slot === targetSlot) {
      let price = prices[index - 1];     // pseudocode starts at 1, JavaScript starts at 0
      return price;
    }
    index = index + 1;
  }

  return -1;
}

let slots  = ["A1", "A2", "B1", "B2"];
let prices = [125, 150, 200, 175];       // cents

let pressed = "B1";
let result = findPrice(slots, prices, pressed);

if (result > 0) {
  console.log("Slot " + pressed + " costs " + result + " cents");
} else {
  console.log("That slot is empty");
}
```

### Python

```python
# CODE_RUNNER: Add a third parallel list for the snack name and return that instead of the price.

def find_price(slots, prices, target_slot):
    index = 1

    for slot in slots:
        if slot == target_slot:
            price = prices[index - 1]      # pseudocode starts at 1, Python starts at 0
            return price
        index = index + 1

    return -1

slots  = ["A1", "A2", "B1", "B2"]
prices = [125, 150, 200, 175]              # cents

pressed = "B1"
result = find_price(slots, prices, pressed)

if result > 0:
    print("Slot " + pressed + " costs " + str(result) + " cents")
else:
    print("That slot is empty")
```

## Side by Side Cheat Sheet

| Idea | Pseudocode | JavaScript | Python |
|------|-----------|------------|--------|
| Assignment | `x <- 5` | `let x = 5;` | `x = 5` |
| Output | `DISPLAY(x)` | `console.log(x)` | `print(x)` |
| Input | `INPUT("msg")` | `prompt("msg")` | `input("msg")` |
| First list position | `1` | `0` | `0` |
| Procedure | `PROCEDURE f(a)` | `function f(a) {}` | `def f(a):` |
| Block markers | `{ }` | `{ }` | indentation |
| Equality test | `=` | `===` | `==` |
| Not equal | `NOT (a = b)` | `a !== b` | `a != b` |
| Add to the end | `APPEND(l, v)` | `l.push(v)` | `l.append(v)` |
| Insert at position i | `INSERT(l, i, v)` | `l.splice(i-1, 0, v)` | `l.insert(i-1, v)` |
| Remove at position i | `REMOVE(l, i)` | `l.splice(i-1, 1)` | `l.pop(i-1)` |
| Length | `LENGTH(l)` | `l.length` | `len(l)` |
| Loop over a list | `FOR EACH x IN l` | `for (let x of l)` | `for x in l:` |
| Loop until | `REPEAT UNTIL (c)` | `while (!c) {}` | `while not c:` |
| Text from a number | automatic | automatic | needs `str(x)` |
| Printing a whole list | `DISPLAY(l)` | `Group A,Group B` | `['Group A', 'Group B']` |

**The three things that trip people up most:**

1. Python will not glue a number onto a string with `+`. Wrap it in `str()` first. JavaScript converts it
   for you without asking.
2. Python has no curly braces. The indentation is the block. Line up your spaces or the code breaks.
3. The two languages print a list differently. Run topic 10 in both and compare. JavaScript flattens the
   list into a plain comma separated line. Python shows the brackets and the quote marks. The list itself
   is the same either way, only the printing is different.
