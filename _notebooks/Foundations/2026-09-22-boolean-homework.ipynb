## 3.05 Boolean Expressions

**Homework: Popcorn Hack 4, SFI Backend Validator**

**Task from the lesson:** build a validator for candidate SFI automotive records. A record is valid when it has a non empty product name, an accepted racing category, a non empty spec number, an effective date, and a spec number that is not already in the backend. The runner has to print whether each record is accepted or rejected.

**How I solved it:** each rule becomes one Boolean variable, and the record is valid only when all of them are true and the duplicate check is false. `and` joins the rules, and `not` flips the duplicate check. The same Boolean values are reused to build the reason list, so a rejected record says why.

| Rule | Boolean expression |
| --- | --- |
| Product name is filled in | `is_filled(record["product_name"])` |
| Category is a racing category | `record["category"] in valid_categories` |
| Spec number is filled in | `is_filled(record["spec_number"])` |
| Effective date is filled in | `is_filled(record["effective_date"])` |
| Spec number is new | `not duplicate_spec` |

```python
# CODE_RUNNER: Run it, then add your own record to test_records and predict accepted or rejected.
existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    }
]

def is_filled(value):
    return value != "" and value is not None

def validate_record(record):
    has_product_name = is_filled(record["product_name"])
    valid_category = record["category"] in valid_categories
    has_spec_number = is_filled(record["spec_number"])
    has_effective_date = is_filled(record["effective_date"])
    duplicate_spec = record["spec_number"] in existing_spec_numbers

    record_is_valid = (
        has_product_name
        and valid_category
        and has_spec_number
        and has_effective_date
        and not duplicate_spec
    )

    reasons = []
    if not has_product_name:
        reasons.append("missing product name")
    if not valid_category:
        reasons.append("category is not a racing category")
    if not has_spec_number:
        reasons.append("missing spec number")
    if not has_effective_date:
        reasons.append("missing effective date")
    if duplicate_spec:
        reasons.append("spec number already in the backend")

    return record_is_valid, reasons

for record in test_records:
    record_is_valid, reasons = validate_record(record)
    print("Record: " + record["product_name"] + " (spec " + record["spec_number"] + ")")
    print("  Category valid: " + str(record["category"] in valid_categories))
    print("  Duplicate spec: " + str(record["spec_number"] in existing_spec_numbers))
    if record_is_valid:
        print("  RESULT: ACCEPTED, record created")
    else:
        print("  RESULT: REJECTED, " + ", ".join(reasons))
```

### Output and What It Shows

| Record | Spec | Result | Why |
| --- | --- | --- | --- |
| Multiple Disc Clutch Assemblies | 1.2 | ACCEPTED | every rule is true and the spec is new |
| Replacement Flywheels | 3.1 | REJECTED | Street Car is not in `valid_categories` |
| Existing Flywheel Record | 1.1 | REJECTED | spec 1.1 is already in the backend |

The three records are picked on purpose. One passes, one fails a category check, and one fails only the duplicate check. That proves the `and` chain is really testing every part, not just the first one.
